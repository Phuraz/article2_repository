```python
import numpy as np
import emcee
import harmonic as hm

# Define the prior probability density function (PDF)
def ln_prior_pdf(theta):
    if -5 < theta[0] < 5 and -5 < theta[1] < 5:
        return 0.0
    return -np.inf

# Define the log-likelihood function
def ln_likelihood(theta, mean, cov):
    return -0.5 * np.sum((theta - mean) ** 2 / np.diag(cov))

# Define the log-posterior function (unnormalized)
def ln_posterior_pdf(theta, mean, cov):
    lp = ln_prior_pdf(theta)
    if not np.isfinite(lp):
        return -np.inf
    return lp + ln_likelihood(theta, mean, cov)
```

```python
import arviz as az
# Set up a sampler (e.g., emcee)
ndim, nwalkers = 2, 32
mean = np.array([0, 0])
cov = np.diag([1, 1])

# Initial positions for walkers
p0 = np.random.rand(nwalkers, ndim)
sampler = emcee.EnsembleSampler(nwalkers, ndim, ln_posterior_pdf, args=[mean, cov])

# Sample the posterior distribution
sampler.run_mcmc(p0, 1000)

# Get the posterior chains
chains = az.from_emcee(sampler)
```

```python
# Instantiate harmonic's evidence class
# Pass the MCMC chains and the log-prior/log-likelihood functions
ev = hm.Evidence(
    chains=chains,
    log_prior_fn=ln_prior_pdf,
    log_likelihood_fn=lambda theta: ln_likelihood(theta, mean, cov)
)

# Train the normalizing flow model
# 'temperature' is a parameter that controls the concentration of the flow
# Recommended temperatures are in the range (0.0, 1.0)
# 'model' can be set to 'rqspline' (rational quadratic spline) for more complex posteriors
ev.train_flow(temperature=0.8, model='rqspline')

# Compute the learned harmonic mean estimate of the evidence
ln_evidence, ln_evidence_errors = ev.compute_ln_evidence()

# Print the result
print(f"Learned harmonic mean estimate of log-evidence: {ln_evidence} ± {ln_evidence_errors}")
```
